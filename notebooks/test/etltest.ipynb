```python
from pathlib import Path
import pandas as pd
import numpy as np
import re
import pyarrow
```

```python
RAW_PATH = Path("../data/raw")

dfs = {}

for file in RAW_PATH.glob("*.csv"):
    dfs[file.stem] = pd.read_csv(file)
```

```python
for name, df in dfs.items():
    print("=" * 70)
    print(name.upper())
    print("=" * 70)

    print(df.info())

    print("\nMissing values")
    print(df.isna().sum())

    print("\nDuplicate rows:", df.duplicated().sum())

    print("\nFirst rows")
    display(df.head())
```

```python
for name in dfs:
    dfs[name] = dfs[name].replace(r"\N", np.nan)
```

```python
def to_ms(value):
    if pd.isna(value):
        return pd.NA

    v = str(value).strip()

    # \N or empty
    if v in {"\\N", ""}:
        return pd.NA

    # +1 Lap / +2 Laps → not numeric
    if "Lap" in v:
        return pd.NA

    # +5.291s
    m = re.fullmatch(r"\+([\d.]+)s", v)
    if m:
        return int(float(m.group(1)) * 1000)

    # mm:ss.SSS or mm:ss.SS or mm:ss
    m = re.fullmatch(r"(\d+):(\d{1,2})(?:\.(\d+))?", v)
    if m:
        minutes = int(m.group(1))
        seconds = int(m.group(2))
        frac = m.group(3)

        ms = int(frac.ljust(3, "0")[:3]) if frac else 0
        return (minutes * 60 + seconds) * 1000 + ms

    return pd.NA
```

```python
TIME_COLUMNS = {
    "lap_times": ["time"],
    "pit_stops": ["duration"],
    "qualifying": ["q1", "q2", "q3"],
    "results": ["time"]
}

for name, df in dfs.items():
    if name in TIME_COLUMNS:
        for col in TIME_COLUMNS[name]:
            if col in df.columns:
                df[col] = df[col].apply(to_ms).astype("Int64")
```

```python
dfs["races"]["time"] = pd.to_datetime(
    dfs["races"]["time"],
    format="%H:%M:%S",
    errors="coerce"
).dt.time
```

```python
for name, df in dfs.items():
    print("\n", name)
    print(df.dtypes[df.dtypes == "object"])
```

```python
def parse_gap(value):
    if pd.isna(value):
        return pd.Series([pd.NA, pd.NA, pd.NA])

    v = str(value).strip()

    # +5.291s → time gap
    m = re.fullmatch(r"\+([\d.]+)s", v)
    if m:
        return pd.Series([int(float(m.group(1)) * 1000), pd.NA, "time"])

    # +1 Lap / +2 Laps → lap gap
    m = re.fullmatch(r"\+(\d+)\s*Laps?", v)
    if m:
        return pd.Series([pd.NA, int(m.group(1)), "lap"])

    # DNF / DNS / DSQ etc
    return pd.Series([pd.NA, pd.NA, v])
```

```python
dfs["results"][["gap_ms", "gap_laps", "gap_type"]] = dfs["results"]["time"].apply(parse_gap)
```

```python
for name, df in dfs.items():
    print("=" * 70)
    print(name.upper())
    print("=" * 70)

    print(df.info())

    print("\nMissing values")
    print(df.isna().sum())

    print("\nDuplicate rows:", df.duplicated().sum())

    print("\nFirst rows")
    display(df.head())
```

```python
OUTPUT = Path("../data/processed/clean_parquet")
OUTPUT.mkdir(parents=True, exist_ok=True)

for name, df in dfs.items():
    df.to_parquet(
        OUTPUT / f"{name}.parquet",
        index=False
    )
```
